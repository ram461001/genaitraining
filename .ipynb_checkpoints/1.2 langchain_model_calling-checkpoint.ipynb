# LangChain Model Calling
This notebook shows how to call LLMs using LangChain with simple examples.

```python
pip install langchain langchain-ollama langchain-openai
```

## 1. Call a Local Model via Ollama

```python
from langchain_ollama import ChatOllama

llm = ChatOllama(model='llama3.2')

response = llm.invoke('What is the capital of France?')
print(response.content)
```

## 2. Call OpenAI Model (optional)
Requires an OpenAI API key.

```python
import os
from langchain_openai import ChatOpenAI

os.environ['OPENAI_API_KEY'] = 'your-api-key-here'

llm = ChatOpenAI(model='gpt-4o-mini')

response = llm.invoke('What is the capital of France?')
print(response.content)
```

## 3. System + User Message

```python
from langchain_ollama import ChatOllama
from langchain_core.messages import SystemMessage, HumanMessage

llm = ChatOllama(model='llama3.2')

messages = [
    SystemMessage(content='You are a helpful assistant that replies in one sentence.'),
    HumanMessage(content='Explain what Python is.')
]

response = llm.invoke(messages)
print(response.content)
```

## 4. Prompt Template
Prompt templates let you reuse prompts with different inputs.

```python
from langchain_ollama import ChatOllama
from langchain_core.prompts import ChatPromptTemplate

llm = ChatOllama(model='llama3.2')

prompt = ChatPromptTemplate.from_messages([
    ('system', 'You are an expert in {topic}.'),
    ('human', '{question}')
])

chain = prompt | llm

response = chain.invoke({
    'topic': 'machine learning',
    'question': 'What is overfitting?'
})

print(response.content)
```

## 5. Streaming Response

```python
from langchain_ollama import ChatOllama

llm = ChatOllama(model='llama3.2')

for chunk in llm.stream('Tell me a short joke.'):
    print(chunk.content, end='', flush=True)
```

## Summary

| Concept | Description |
|---|---|
| `ChatOllama` | Run models locally via Ollama |
| `ChatOpenAI` | Call OpenAI models via API |
| `invoke()` | Send a prompt and get a full response |
| `stream()` | Get response token by token |
| `ChatPromptTemplate` | Reusable prompt with variables |
| `chain = prompt | llm` | LangChain pipe to connect prompt → model |

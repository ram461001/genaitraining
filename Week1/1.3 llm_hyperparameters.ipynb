# LLM Hyperparameters
Key parameters that control how an LLM generates text.

```python
from langchain_ollama import ChatOllama
```

## 1. Temperature
Controls **randomness** of the output.

- `0.0` → deterministic, always picks the most likely word (factual tasks)
- `1.0` → balanced
- `2.0` → very random and creative (may be incoherent)

**Rule of thumb:** Use low temperature for facts, high for creative writing.

```python
prompt = 'Give me a one-line tagline for a coffee shop.'

for temp in [0.0, 0.7, 1.5]:
    llm = ChatOllama(model='llama3.2', temperature=temp)
    response = llm.invoke(prompt)
    print(f'temperature={temp} -> {response.content}')
```

```python
for temp in [0.0, 0.7, 1.5]:
    llm = ChatOllama(model='llama3.2', temperature=temp)
    response = llm.invoke(prompt)
    print(f'temperature={temp} -> {response.content}')
```

## 2. Top-P (Nucleus Sampling)
Controls **vocabulary diversity** by limiting which words the model considers.

- `top_p=0.1` → only picks from the top 10% most likely words (focused)
- `top_p=0.9` → picks from a wider pool (more varied)
- `top_p=1.0` → consider all words (default)

**Tip:** Tune either `temperature` OR `top_p`, not both at once.

```python
prompt = 'Describe the ocean in one sentence.'

for top_p in [0.1, 0.5, 1.0]:
    llm = ChatOllama(model='llama3.2', top_p=top_p)
    response = llm.invoke(prompt)
    print(f'top_p={top_p} -> {response.content}')
```

## 3. Max Tokens
Limits how long the response can be.

```python
prompt = 'Explain machine learning.'

for max_tokens in [20, 100]:
    llm = ChatOllama(model='llama3.2', num_predict=max_tokens)
    response = llm.invoke(prompt)
    print(f'max_tokens={max_tokens} -> {response.content}')
    print()
```

## 4. Top-K
Limits the model to pick from the **top K most likely tokens** at each step.

- `top_k=1` → always picks the single most likely word (greedy)
- `top_k=50` → picks from top 50 candidates

```python
prompt = 'Write a creative name for a robot.'

for top_k in [1, 50]:
    llm = ChatOllama(model='llama3.2', top_k=top_k)
    response = llm.invoke(prompt)
    print(f'top_k={top_k} -> {response.content}')
```

## Summary

| Parameter | What it controls | Low value | High value |
|---|---|---|---|
| `temperature` | Randomness | Focused, deterministic | Creative, random |
| `top_p` | Vocabulary pool size | Narrow word choices | Wider word choices |
| `top_k` | Top candidates per step | Greedy/predictable | More variety |
| `max_tokens` | Response length | Short answer | Longer answer |

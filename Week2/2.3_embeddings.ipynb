# 2.3 — Embeddings

An **embedding** converts text into a list of numbers (a vector) that captures its **meaning**.

- Similar sentences → similar vectors (close together in space)
- Different sentences → different vectors (far apart)

This is how we do **semantic search** — finding documents by meaning, not just keywords.

```
'I love pizza'  →  [0.12, -0.45, 0.88, ...]
'Pizza is my favorite food'  →  [0.10, -0.47, 0.85, ...]  ← very close!
'The stock market crashed'   →  [-0.71, 0.33, -0.12, ...]  ← far away
```

## 1. Create Embeddings with Ollama (Local, No API Key)

````python
from langchain_ollama import OllamaEmbeddings

embeddings = OllamaEmbeddings(model='llama3.2')

text = 'I love eating pizza on weekends'
vector = embeddings.embed_query(text)

print(f'Text: "{text}"')
print(f'Vector length: {len(vector)}')
print(f'First 5 values: {vector[:5]}')
````

## 2. Embed Multiple Documents

````python
sentences = [
    'Employees get 20 days of annual leave.',
    'Annual leave is 20 days for all staff.',
    'The office serves free lunch every day.',
    'Remote work is allowed 3 days per week.',
]

vectors = embeddings.embed_documents(sentences)

print(f'Embedded {len(vectors)} sentences')
print(f'Each vector has {len(vectors[0])} dimensions')
````

## 3. Measuring Similarity (Cosine Similarity)

Cosine similarity measures how close two vectors are:
- `1.0` = identical meaning
- `0.0` = unrelated
- `-1.0` = opposite meaning

````python
import numpy as np

def cosine_similarity(v1, v2):
    v1, v2 = np.array(v1), np.array(v2)
    return np.dot(v1, v2) / (np.linalg.norm(v1) * np.linalg.norm(v2))

query = 'How many vacation days do employees get?'
query_vector = embeddings.embed_query(query)

print(f'Query: "{query}"\n')
for sentence, vec in zip(sentences, vectors):
    score = cosine_similarity(query_vector, vec)
    print(f'  Score: {score:.4f}  |  "{sentence}"')
````

## 4. Key Insight — Semantic vs Keyword Search

````python
# These sentences are semantically similar but share NO keywords
similar_pairs = [
    ('I need to take time off work', 'Request for annual leave'),
    ('The employee is working from home', 'Remote work policy'),
    ('Company perks and benefits', 'Health insurance and gym membership'),
]

print('Semantic similarity between related pairs:\n')
for s1, s2 in similar_pairs:
    v1 = embeddings.embed_query(s1)
    v2 = embeddings.embed_query(s2)
    score = cosine_similarity(v1, v2)
    print(f'  "{s1}"')
    print(f'  "{s2}"')
    print(f'  Similarity: {score:.4f}\n')
````

## Summary

| Concept | Description |
|---------|-------------|
| `embed_query()` | Embed a single query string |
| `embed_documents()` | Embed a list of document strings |
| Cosine similarity | Measures how similar two vectors are (0 to 1) |
| Semantic search | Find documents by meaning, not exact words |
| `OllamaEmbeddings` | Local embeddings — no API key needed |

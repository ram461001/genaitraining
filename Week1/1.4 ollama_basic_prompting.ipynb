# Basic Ollama Model Prompting
This notebook shows how to send prompts to a local Ollama model.

## Prerequisites
- [Ollama](https://ollama.com) installed and running
- A model pulled, e.g. `ollama pull llama3.2`
- Install the Python library: `pip install ollama`

```python
!pip install ollama
```

## 1. Simple Prompt

```python
import ollama

response = ollama.chat(
    model='gemma3:4b',
    messages=[
        {'role': 'user', 'content': 'What is the capital of France?'}
    ]
)

print(response['message']['content'])
```

## 2. System Prompt + User Prompt

```python
response = ollama.chat(
    model='gemma3:4b',
    messages=[
        {'role': 'system', 'content': 'You are a helpful assistant that replies in one sentence.'},
        {'role': 'user', 'content': 'Explain what Python is.'}
    ]
)

print(response['message']['content'])
```

## 3. Streaming Response

```python
stream = ollama.chat(
    model='llama3.2',
    messages=[
        {'role': 'user', 'content': 'Tell me a short joke.'}
    ],
    stream=True
)



for chunk in stream:
    print(chunk['message']['content'], end='', flush=True)
```

## 4. List Available Models

```python
models = ollama.list()
for model in models['models']:
    print(model['model'])
```
